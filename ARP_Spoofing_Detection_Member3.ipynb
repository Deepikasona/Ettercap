# Member 3 – ARP Spoofing Detection

This notebook demonstrates basic gateway MAC comparison and simulated suspicious-activity detection.

**Important limitation:** Google Colab cannot directly read the local Windows ARP table. The network details below were manually recorded from Windows, and the changed MAC address is simulated. This notebook is not live ARP monitoring or packet capture.


```python
# Step 1 - Record network details collected from the Windows laptop
interface = "Wi-Fi"
laptop_ip = "172.20.10.2"
gateway_ip = "172.20.10.1"
trusted_gateway_mac = "C6-C1-7D-4A-BE-64"
source_note = "Manually recorded from Windows ARP neighbour table"

print("Interface:", interface)
print("Laptop IP:", laptop_ip)
print("Gateway IP:", gateway_ip)
print("Trusted gateway MAC:", trusted_gateway_mac)
print("Source:", source_note)
```

```python
# Step 2 - Normal network test (same MAC address)
current_mac = "C6-C1-7D-4A-BE-64"

print("===== NORMAL NETWORK TEST =====")
print("Trusted MAC:", trusted_gateway_mac)
print("Current MAC:", current_mac)

if current_mac.upper() == trusted_gateway_mac.upper():
    print("STATUS: NETWORK NORMAL")
    print("Gateway MAC matches the trusted baseline.")
else:
    print("WARNING: SUSPICIOUS ACTIVITY DETECTED!")
    print("Gateway MAC address mismatch.")
print("================================")
```

```python
# Step 3 - Simulated suspicious gateway test (different MAC address)
current_mac = "99-88-77-66-55-44"  # Simulated value for testing only

print("===== SECURITY ALERT TEST =====")
print("Trusted MAC:", trusted_gateway_mac)
print("Current MAC:", current_mac)

if current_mac.upper() != trusted_gateway_mac.upper():
    print("\nWARNING: SUSPICIOUS ACTIVITY DETECTED!")
    print("Gateway MAC address mismatch.")
    print("Action: Investigate the network.")
else:
    print("\nSTATUS: NETWORK NORMAL")
print("===============================")
```

```python
# Step 4 - Final security report (simulated changed MAC)
current_mac = "99-88-77-66-55-44"  # Simulated value for testing only
alerts = []
if current_mac.upper() != trusted_gateway_mac.upper():
    alerts.append("Gateway MAC address mismatch")

print("========== FINAL SECURITY REPORT ==========")
print("Network Interface:", interface)
print("Laptop IP:", laptop_ip)
print("Gateway IP:", gateway_ip)
print("Trusted Gateway MAC:", trusted_gateway_mac)
print("Current Gateway MAC:", current_mac)
if alerts:
    print("\nSecurity Status: WARNING")
    print("Total Alerts:", len(alerts))
    for alert in alerts:
        print("ALERT:", alert)
else:
    print("\nSecurity Status: NORMAL")
    print("Total Alerts: 0")
print("===========================================")
```
